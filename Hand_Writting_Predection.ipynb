```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
from sklearn.model_selection import train_test_split
```

```python
import tensorflow
from tensorflow import keras
from tensorflow.keras import Sequential
from tensorflow.keras.layers import Dense,Flatten
```

```python
## Keras -> Mnist dataser
```

```python
(X_train,y_train),(X_test,y_test) = keras.datasets.mnist.load_data()
X_train.shape
```

```python
X_train
```

```python
## 0-1 modde nia asbo every datapoint
```

```python
X_train = X_train / 255
X_test = X_test / 255
```

```python
X_train[0]
```

```python
## Now creating ANN
```

```python
model = Sequential()
model.add(Flatten(input_shape=(28,28)))
model.add(Dense(128,activation="relu"))
model.add(Dense(10,activation="softmax"))
```

```python
model.summary()
```

```python
model.compile(loss="sparse_categorical_crossentropy",optimizer="Adam",metrics=["accuracy"])
```

```python
ann_history = model.fit(X_train,y_train,epochs=10,validation_split=0.2)
```

```python
y_pro = model.predict(X_test)
y_predict = y_pro.argmax(axis=1)
```

```python
from sklearn.metrics import accuracy_score
```

```python
print("ANN Model Accurecy Score:- ",accuracy_score(y_test,y_predict)*100)
```

```python
plt.plot(ann_history.history["loss"],label="Train Loss")
plt.plot(ann_history.history["val_loss"],label="Validation Loss")
plt.legend()
plt.show()
```

```python
plt.plot(ann_history.history["accuracy"],label="Train Accuracy")
plt.plot(ann_history.history["val_accuracy"],label="Validation Accuracy")
plt.legend()
plt.show()
```
